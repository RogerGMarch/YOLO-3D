```python
import pandas as pd
video = 'Suspiro'
# Load the parquet file
file_path = f'/media/M2_disk/roger/sonar/YOLO-3D/videos/out/{video}_tracking_data.parquet'
tracking_df = pd.read_parquet(file_path)

# Split the dataframe based on keypoint_index
keypoint_negative_df = tracking_df[tracking_df.keypoint_index == -1]
keypoint_positive_df = tracking_df[tracking_df.keypoint_index >= 0]
columns_to_remove = [
    'spatial_spread',
    'avg_distance',
    'formation_area',
    'symmetry_score',
    'num_performers',
    'distance'
]

keypoint_positive_df = keypoint_positive_df.drop(columns=columns_to_remove, errors='ignore')

# Define output file paths
negative_output_path = f'/media/M2_disk/roger/sonar/YOLO-3D/videos/out/{video}_company_metrics.csv'
positive_output_path = f'/media/M2_disk/roger/sonar/YOLO-3D/videos/out/{video}_tracking_data.csv'

# Save to CSV
keypoint_negative_df.to_csv(negative_output_path, index=False)
keypoint_positive_df.to_csv(positive_output_path, index=False)

# Print information about the split
print(f"Original dataframe: {tracking_df.shape[0]} rows")
print(f"Keypoint index -1: {keypoint_negative_df.shape[0]} rows saved to {negative_output_path}")
print(f"Keypoint index ≥0: {keypoint_positive_df.shape[0]} rows saved to {positive_output_path}")
```
